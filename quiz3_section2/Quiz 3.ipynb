(2) Aggregate and query

<font size = "3">

Import the data from "sprint_results.csv" and assign it to a DataFrame. Do the following:

- Extract the subset corresponding to 'constructorId == 51' using ```.query()```
- Group by the column "driverId", and then aggregate the maximum over the "laps" column for each driver.
- Extract the subset of drivers with maximum of laps $\ge 20$. Assign this to a DataFrame with a name of your choice.

For full credit, perform all the steps in bullet points above in a **single line** using chaining. (If you use line breaks or wrap in parentheses it doesn't literally need to be one line).

**Hint:** I only see your final submission, not the code you wrote along the way. It might be easier to break it up into multiple lines, then combine when you're done.

````python
# Your answer here

sprinting = pd.read_csv("data_raw/sprint_results.csv")

many_laps = (sprinting.query("constructorId == 51")
                      .groupby(by = "driverId")
                      .agg(max_laps = ("laps", "max"))
                      .query("max_laps >= 20"))

print(many_laps)
````

(3) Recode a numeric column

<font size = "3">

- Import data from "races.csv" and assign it to a DataFrame
- Recode the "year" column into a new column "year_brackets"
with the following categories

$\qquad$ ``` ["1950-1965","1966-1980","1981-onwards"] ```

The easiest method is ``pd.cut()`` but you are not required to use this.

````python
# Your answer here

races = pd.read_csv("data_raw/races.csv")

year_bins = [1950, 1965, 1980, float("inf")]
year_labels = ["1950-1965","1966-1980","1981-onwards"]

races["year_brackets"] = pd.cut(races["year"], bins=year_bins, right=True, labels=year_labels)
print(races["year_brackets"])
````

(4) Aggregate and sort

<font size = "3">

Import the data from "sprint_results.csv" and assign it to a DataFrame (or use the same DataFrame you created in problem (2)). Do the following:

- Group by "driverId", and then aggregate the average (mean) over the "laps" column.
- Sort the drivers by average laps in **descending** order using ``.sort_values()``. Assign this to a DataFrame with a name of your choice.

For full credit, perform both steps in bullet points above in a **single line** using chaining. (If you use line breaks or wrap in parentheses it doesn't literally need to be one line).

**Hint:** Same hint as problem 2. You can chain commands together later.

````python
# Your answer here

# sprinting = pd.read_csv("data_raw/sprint_results.csv")

driver_mean_laps = (sprinting.groupby(by = "driverId")
                           .agg(mean_laps = ("laps", "mean"))
                           .sort_values(by = "mean_laps", ascending=False))

print(driver_mean_laps)
````

(6) Merge dataset

- Import the data from "sprint_results.csv" and "constructors.csv", assigning them to two DataFrames (you can also re-use the DataFrames you created above.)

- Create a new dataset with  ``pd.merge()``. Treat "sprint_results.csv" as the primary dataset and "constructors.csv" as the secondary dataset. The merge should be based on the column "constructorId".
the column "constructorId"
- To get full credit **ONLY** merge the "nationality" column from the secondary dataset (not all the columns)
- Print (or display) the resulting merged DataFrame to the screen.

**Hint**: Before you start typing "pd.merge(...)" try printing out the two DataFrames you want to merge, excluding any columns that shouldn't be included.

````python
# Your answer here

sprinting = pd.read_csv("data_raw/sprint_results.csv")
constructors = pd.read_csv("data_raw/constructors.csv")

constructors_subset = constructors[["constructorId", "nationality"]]

merged_data = pd.merge(left=sprinting, right=constructors_subset, on ="constructorId", how="left")

print(merged_data)


````
