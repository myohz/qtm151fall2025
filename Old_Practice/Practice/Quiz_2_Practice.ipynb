Question 5:  Given a DataFrame, you will want to randomly assign each row to one of two groups. To do so, you will add a new column to the DataFrame called "random_var" which is filled with random numbers from a given distribution. Then you will create two new DataFrames, based on the whether each row's random variable satisfies a given logical condition. These two DataFrames represent the random assignment into two groups.

```python
num_rows, num_cols = data_frame.shape

random_var = np.random.choice([1, 2], size = num_rows)

data_frame["random_var"] = random_var


group1 = data_frame.query("random_var = 1") 
group2 = data_frame.query("random_var = 2")

# group1 = df[df['random_var'] > 0.5]
# group2 = df[df['random_var'] <= 0.5]
```

Question 6: Continuing question 5, you will make two scatter plots from the data in each randomly assigned group. So that they appear side-by-side, you will need to use the subplots command from the matplotlib.pyplot library.

```python

fig, subfig_array = plt.subplots(nrows=1, ncols=2, figsize = (6,3))

subfig_array[0].scatter(x = group1["x_value"], y = group1["y_value"])
subfig_array[0].set_title("y vs. x for Group 1")
subfig_array[0].set_xlabel("x")
subfig_array[0].set_ylabel("y")

subfig_array[1].scatter(x = group2["x_value"], y = group2["y_value"])
subfig_array[1].set_title("y vs. x for Group 1")
subfig_array[1].set_xlabel("x")
subfig_array[1].set_ylabel("y")

plt.tight_layout()
plt.show()
```
